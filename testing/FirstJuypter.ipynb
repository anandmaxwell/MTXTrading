```python
import pandas as pd
import datetime as dt
import vectorbt as vbt


end_date = dt.datetime.now()
start_date = end_date - dt.timedelta(days=1)


priceData = vbt.YFData.download(
    'SCODATUBES.NS', start=start_date, end=end_date, interval='1m', missing_index='drop' 
).get('Close')
priceData = priceData.rename_axis('Date').reset_index()
priceData['Date'] = pd.to_datetime(priceData['Date'])
priceData = priceData.set_index('Date').sort_index()


print(priceData.head())
print(priceData.tail())


```

```python

rsi = vbt.RSI.run(priceData, window=14, short_name='RSI')



entries = rsi.rsi_crossed_above(70)  # Buy when RSI crosses above 70
exits = rsi.rsi_crossed_below(30)    # Sell when RSI crosses below
portfolio = vbt.Portfolio.from_signals(
    priceData, entries, exits ,freq='1min',
    init_cash=100000, fees=0.001, slippage=0.001

)   

#print(portfolio.returns().to_string()      )
print(portfolio.stats().to_string())
```
