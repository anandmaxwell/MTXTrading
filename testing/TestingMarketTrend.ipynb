```python
from indicators.candlePattern.CandleStickInfoGenerator import CandleStickInfoGenerator

from dataLoader.YahooFinanceDataCollector import YahooFinanceDataCollector

import pandas as pd


#/////////////////////////////////////////
# Input
#/////////////////////////////////////////
ticker ="BHEL.NS"
period = "1d"
interval = "2m"



#...................................................................................
# Need to get only the  price
#...................................................................................
yahooFinanceDataCollector = YahooFinanceDataCollector(ticker, period=period , interval = interval)
tickerDFPrice = yahooFinanceDataCollector.collectDate()

tickerDFPrice = tickerDFPrice.copy()
tickerDFPrice['Date'] = pd.to_datetime(tickerDFPrice.index, format='%Y-%m-%d' , errors='coerce')
tickerDFPrice.set_index('Date', inplace=True)


print(tickerDFPrice)
# print(tickerDFPrice)
```

```python
import math

timeLength = 15
print(len(tickerDFPrice))

pointer = 0

numColumns = len(tickerDFPrice.columns)

tickerDFPrice.insert(numColumns,'support_starrting_candle_heigth', 0)
print("Number of columns in the tickerDFPrice: " + str(numColumns))

numColumns = len(tickerDFPrice.columns)
tickerDFPrice.insert(numColumns,'support_ending_candle_heigth', 0)
print("Number of columns in the tickerDFPrice: " + str(numColumns))

numColumns = len(tickerDFPrice.columns)
tickerDFPrice.insert(numColumns,'resistance_starting_candle_heigth', 0)
print("Number of columns in the tickerDFPrice: " + str(numColumns))

numColumns = len(tickerDFPrice.columns)
tickerDFPrice.insert(numColumns,'resistance_ending_candle_heigth', 0)
print("Number of columns in the tickerDFPrice: " + str(numColumns))


numColumns = len(tickerDFPrice.columns)
tickerDFPrice.insert(numColumns,'x0', 0)
print("Number of columns in the tickerDFPrice: " + str(numColumns))


numColnumColumns = len(tickerDFPrice.columns)
tickerDFPrice.insert(numColumns,'x1', 0)
print("Number of columns in the tickerDFPrice: " + str(numColumns))

numColnumColumns = len(tickerDFPrice.columns)
tickerDFPrice.insert(numColumns,'mType', 0)
print("Number of columns in the tickerDFPrice: " + str(numColumns))




# tickerDFPrice.info

startingPoint = pointer

for princeInfo in tickerDFPrice.itertuples():

    endPoint = startingPoint+timeLength

    if endPoint < len(tickerDFPrice):

        starrting_candle = tickerDFPrice.iloc[startingPoint]
        ending_candle = tickerDFPrice.iloc[endPoint]

        x0 = tickerDFPrice.index[startingPoint]
        x1 = tickerDFPrice.index[endPoint]
        print(f"At time -> {x0}" + f" x0: {x0} " + f" x1: {x1} " + f"currentpoint -> {startingPoint} ")
        
        support_starrting_candle_heigth = starrting_candle['High']
        support_ending_candle_heigth = ending_candle['High']- support_starrting_candle_heigth

        resistance_starting_candle_heigth = starrting_candle['Low']
        resistance_ending_candle_heigth = ending_candle['Low'] - resistance_starting_candle_heigth

        starrting_candle['support_starrting_candle_heigth'] = starrting_candle['High']
        starrting_candle['support_ending_candle_heigth'] = ending_candle['High']
        starrting_candle['resistance_starting_candle_heigth'] = starrting_candle['Low']
        starrting_candle['resistance_ending_candle_heigth'] = ending_candle['Low']
        starrting_candle['x0'] = x0
        starrting_candle['x1'] = x1

        support_angle = math.degrees(math.sin(support_ending_candle_heigth / timeLength))
        resistance_angle = math.degrees(math.sin(resistance_ending_candle_heigth / timeLength))
        print(f"At time -> {x0}" + f"Support Angle: {support_angle} "+ f" Resistance Angle: {resistance_angle}" + f"currentpoint -> {startingPoint} ")
        if support_angle > 0 and resistance_angle > 0:
            starrting_candle['mType'] ="Bullish" 
        elif support_angle < 0 and resistance_angle < 0:
            starrting_candle['mType'] ="Bearish" 
        else:
            starrting_candle['mType'] ="Sideways" 

        tickerDFPrice.iloc[startingPoint] = starrting_candle
        

       

    startingPoint = endPoint
```

```python
 # 5. Create Candlestick + VWAP Plot

import plotly.io as pio
import plotly.graph_objects as go
pio.renderers.default='browser'    
df = tickerDFPrice

fig =  go.Figure(
        layout_yaxis_range=[min(df['Close']), max(df['Close'])]
    )
    



fig.add_trace(go.Candlestick(
        x=df.index,
        open=df['Open'],
        high=df['High'],
        low=df['Low'],
        close=df['Close'],
        name='Candlestick'
    ))

for priceRec in tickerDFPrice.iterrows():

    if priceRec[1]['support_starrting_candle_heigth'] != 0 and priceRec[1]['support_ending_candle_heigth'] != 0:
    
        fig.add_shape(
            type="line",
            x0=priceRec[1]['x0'],
            y0=priceRec[1]['support_starrting_candle_heigth'],
            x1=priceRec[1]['x1'],
            y1=priceRec[1]['support_ending_candle_heigth'],
            line=dict(color="RoyalBlue", width=2)  # Optional: customize line appearance
        )

        fig.add_shape(
            type="line",
            x0=priceRec[1]['x0'],
            y0=priceRec[1]['resistance_starting_candle_heigth'],
            x1=priceRec[1]['x1'],
            y1=priceRec[1]['resistance_ending_candle_heigth'],
            line=dict(color="red", width=2)  # Optional: customize line appearance
        )

        fig.add_trace(go.Scatter(x=priceRec[1]['x0'],
                             y=priceRec[1]['resistance_starting_candle_heigth'] * 2.01, # Adjust position for visibility
                             mode='markers',
                             marker_symbol='triangle-down',
                             marker_color='red',
                             marker_size=10,
                             name='Sell Signal'))




fig.show()
```
